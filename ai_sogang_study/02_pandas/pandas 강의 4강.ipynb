boolean indexing

true/false 배열을 만들어서 이를 이용해서 인덱싱



```python
import pandas as pd
import numpy as np
a1 = [60, 84, 80, 70, "문예부"]
a2 = [77, 62, 95, 85, "과학부"]
a3 = [61, 97, 72, np.nan, "농구부"]
a4 = [75, 65, 95, np.nan, "다도부"]
cols = ["국어", "영어", "수학", "과학", "소속"]
b1 = ["문예부", "과학부", "다도부"]
df = pd.DataFrame([a1, a2, a3, a4], index=list("ABCD"), columns=cols)
```

```python
df
```

```python
df.loc[["A"]]
```

```python
df.loc[:,["영어"]]
```

```python
filter1 = df["영어"] > 80
filter1
```

```python
df.loc[filter1]
```

```python
df.loc[filter1, "국어" : "과학"]
```

```python
filter2 = df["국어"] > 60
filter3 = df["수학"] > 75
df.loc[filter2 & filter3]
```

```python
# 필터4 : 과학 시험을 쳤다. 쳤으면 true 안쳤으면 false
filter4 = df["과학"].isnull()
```

```python
df.loc[filter4]
```

```python
df.loc[~filter4]
```

```python
# 과학 시험을 안쳤더라도 수학 시험이 90점이 넘으면 면제라고 하면? -> 필터5 : 과학 시험을 쳤다 or 수학이 90 넘는다
filter5 = (~df["과학"].isnull()) | (df["수학"] > 90)
filter5
```

```python
df.loc[filter5]
```

```python
df["소속"].isin(b1) # 속하면 true, 속하지 않으면 false
```
