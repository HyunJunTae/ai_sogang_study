astype/to_numercic (to_numeric은 시리즈에만 사용 가능)

astype 으로 타입을 바꿀 수 있음.

근데 astype에서는, 바꿀 수 없는 걸 만나면 어케할지 정할 수 있음.

ex)float로 바꿔주다가, 문자열같은거 만나면 errors='ignore'로 설정해서 안바꾸고 무시하고 넘어가게 할 수 있음.

df['C'].astype('float', errors='ignore')

이렇게 하면 C열은 그냥 넘어감.

근데 C열에 문자열이랑 숫자랑 섞여있을 수도 있잖아?

-> 그럴 때 쓰는게 to_numeric

errors='coerce' -> 문자열같은 문자 만나면 Nan으로 바꿈.

fillna로 다시 채우주면 됨.
